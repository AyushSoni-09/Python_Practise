Load 'auto_mpg.csv' data set and answer the below given operations programmatically using Pandas,

What is the total number of data points (including missing values) in the auto_mpg dataset?
For the 'horsepower' column, how many non-missing values are there?  How many missing values are there?
What are the counts of each unique value in the 'origin' column?  Which origin is the most frequent?
What is the distribution of car manufacturers in the dataset? (using value_counts on the name column after cleaning it)

```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_csv('auto_mpg.csv')
df
```

```python
data_size = df.size
print("Total number of Element in dataset: ",data_size)
```

```python
Non_missing_values= df['horsepower'].count()
missing_values = df['horsepower'].isnull().sum()
print("Total number of Non-missing values in horsepower : ",Non_missing_values)
print("Total number of Missing values in horsepower :", missing_values)
```

```python
# What are the counts of each unique value in the 'origin' column?  Which origin is the most frequent?
origin_unique_count = df['origin'].value_counts()
print("Unique value count in origin column: ",origin_unique_count)

most_frequent_origin = origin_unique_count.idxmax()
print("Most frequent value in origin colmun : ", most_frequent_origin)
```

```python
df=df.dropna(subset=['name'])
df['manufacture'] = df['name'].str.split().str[0]
distribution_car_man = df['manufacture'].value_counts()
print("Distribution of car manufacturer in dataset: ",distribution_car_man)
frequent_manufacturer = distribution_car_man.idxmax()
print("most frequent manufacturer in dataset: ",frequent_manufacturer)
```
