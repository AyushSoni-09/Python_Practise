join() Function

Purpose:

Used to combine two DataFrames based on the index instead of columns.
Efficient when working with indexed data.
Usage:

When merging DataFrames based on index values.
When quickly joining hierarchical or time-series data.

Syntax:

df1.join(df2, how='left', on=None)

```python
import pandas as pd
import numpy as np
```

```python
# Sample DataFrames
df1 = pd.DataFrame({'key': ['A', 'B', 'C', 'D'], 'col1': [1, 2, 3, 4]})
df2 = pd.DataFrame({'key': ['B', 'C', 'D', 'E'], 'col2': [5, 6, 7, 8]})
print("Data Frame -1 \n", df1)
print("Data Frame -2\n",df2)
```

```python
#1. Inner Merge
inner_join = pd.merge(df1,df2,on='key', how='inner')
inner_join
```

```python
##2. Outer Merge
outer_merge = pd.merge(df1,df2,on='key',how='outer')
outer_merge
```

```python
# left merge
left_merge= pd.merge(df1,df2,on='key',how='left')
left_merge
```

```python
#right merge
right_merge = pd.merge(df1,df2,on='key',how='right')
right_merge
```

```python
# --- Join Operations (assuming 'key' is the index in both DataFrames) ---
df_1=df1.set_index('key')
df_2= df2.set_index('key')
print("Data Frame -1 \n", df_1)
print("Data Frame -2\n",df_2)
```

```python
#1. inner join 
inner_join = df_1.join(df_2,how='inner')
inner_join
```

```python
# 2.outer join
outer_join = df_1.join(df_2,how='outer')
outer_join
```

```python
#3.left join
left_join = df_1.join(df_2,how='left')
left_join
```

```python
#4.right join 
right_join = df_1.join(df_2,how='right')
right_join
```
