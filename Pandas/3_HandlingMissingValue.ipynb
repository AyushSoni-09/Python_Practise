```python
import numpy as np
import pandas as pd
```

```python
# Creating a synthetic dataset with different types of missing values
data = {
    'A': [1, 2, np.nan, 4, 9999, 6, '?', 8, 9],
    'B': [5, '?', 7, 8, 9, np.nan, 11, 9999, 13],
    'C': [9999, 15, 16, '?', 18, np.nan, 20, 21, 22]
}
```

```python
#Converting the data set into dataframe
df = pd.DataFrame(data)
```

```python
print("Original Data: ")
df
```

```python
# Replace '?' and 9999 with NaN for standard handling
df.replace(['?',9999], np.nan , inplace=True)
print(df)
df.dtypes
```

```python
# Check for missing values
print("\n Missing Values Count: ")
print(df.isnull().sum())
```

```python
#Droppoing missing value
df.dropna()
```

```python
#Imputing Missing Values: Uses .fillna(df.mean()) to replace NaN with the mean of each column.
df = df.apply(pd.to_numeric , errors='coerce')  # Convert all columns to numeric, coercing errors to NaN
df_impute = df.fillna(df.mean(numeric_only=True))
print(df_impute)
```
