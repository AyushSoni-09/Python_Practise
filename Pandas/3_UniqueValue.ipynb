```python
import pandas as pd 
import numpy as np
```

```python
data = pd.read_csv('auto_mpg.csv')
print("\n. unique data:")
unique_data= data['origin'].unique()
print(unique_data)
```

```python
#Extracts all the distinct values present in the all the columns of the auto_mpg.csv dataset.

for cols in data.columns:
    unique_cols=data[cols].unique()
    print(f"Column: {cols} Unique Values: {unique_cols}")
    print("-"*50)

```
